Exercice 1 : Perceptron simple

Etape 1 : Importer les librairies

```python
import numpy as np
import pandas as pd
```

Etape 2 : Définir les données d'entraînement

```python
X = np.array([[0,0],[0,1],[1,0],[1,1]])
yAnd = np.array([0,0,0,1])
```

```python
df_X = pd.DataFrame(X, columns=['Feature_1', 'Feature_2'])
print("DataFrame X :")
df_X
```

```python
# Création d'un dataframe pour yAnd
df_yAnd = pd.DataFrame(yAnd, columns=['Target'])
print("\nDataFrame yAnd :")
df_yAnd
```

Etape 3 : Initialiser les poids et le biais

```python
np.random.seed(42)  # Pour la reproductibilité des résultats
W = np.random.rand(2)  # Poids aléatoires pour les deux caractéristiques
b = np.random.rand(1)  # Biais aléatoire
print("Poids initiaux :", W)
print("Biais initial :", b)
```

Etape 4 : Définir la fonction d'activation

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

Etape 5 : Définir la prédiction du perceptron

```python
def predict(X, W, b):
    z = np.dot(X, W) + b
    return sigmoid(z)
```

Etape 6 : Définir la fonction de perte

```python
def loss(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
```

Etape 7 : Entraînement du perceptron (descente de gradient)

```python
learning_rate = 0.1
epochs = 2000

for i in range(epochs):
    y_pred = predict(X, W, b)
    # Cacul du gradient
    dW = -2 * np.dot((yAnd - y_pred) * y_pred * (1 - y_pred), X) / len(X)
    db = -2 * np.mean((yAnd - y_pred) * y_pred * (1 - y_pred))
    # Mise à jour des poids et du biais
    W -= learning_rate * dW
    b -= learning_rate * db
    if i % 100 == 0:
        print(f"Epoch {i}, Loss: {loss(yAnd, y_pred)}")
```

Etape 8 : Tester le perceptron

```python
y_pred_final = predict(X, W, b)
print("Prédictions finales :", y_pred_final)
print("Classifications finales :", (y_pred_final > 0.5).astype(int))
```

Même chose mais avec l'entropie croisée binaire comme fonction de perte

```python
np.random.seed(42)  # Pour la reproductibilité des résultats
W = np.random.randn(2)
b = np.random.randn()

def bce(y_true, y_pred, eps=1e-12):
    y_pred = np.clip(y_pred, eps, 1 - eps) # Pour éviter le log(0)
    return -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))

learning_rate = 0.5
epochs = 2000
for i in range(epochs):
    y_pred = predict(X, W, b)
    dz = y_pred - yAnd
    # Calcul du gradient pour la BCE
    dW = np.dot(dz, X) / len(X)
    db = np.mean(dz)
    # Mise à jour des poids et du biais
    W -= learning_rate * dW
    b -= learning_rate * db
    if i % 400 == 0:
        print(f"Epoch {i}, BCE Loss: {bce(yAnd, y_pred)}")

print("Prédictions finales après entraînement avec BCE :", predict(X, W, b).round(3))
```
