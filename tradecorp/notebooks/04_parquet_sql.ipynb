Somme des tailles des fichiers CSV : 159 Ko, taille du répertoire parquet orders_enriched : 88 Ko, parquet est plus économe car :
* il stocke en binaire et pas en ASCII comme CSV, ce qui est plus efficace
* il utilise des algorithmes de compression
* il stocke par colonne qui correspond à un type, le stockage d'un type numérique est plus efficace que son équivalent en chaine de caractères

```python
url = "jdbc:postgresql://postgres:5432/"+os.getenv("POSTGRES_DB")
properties = {
    "user": os.getenv("POSTGRES_USER"),
    "password": os.getenv("POSTGRES_PASSWORD"),
    "driver": "org.postgresql.Driver"
}
```

```python
df_orders_enriched.write.jdbc(url=url,table="public.orders_enriched",mode="append",properties=properties)
```
