La lazy evaluation (évaluation paresseuse) est le principe suivant : Spark n’exécute pas immédiatement les opérations qu'on lui demande. Il construit d’abord un plan d’exécution, puis ne lance réellement les calculs que lorsqu’une action est appelée.

Par exemple, df_min_freight = df_orders.select(min("freight")) est une transformation, elle n'est pas exécutée immédiatement mais est intégrée dans un DAG (graphe de calcul). Ce n'est que lorsqu'une action du type df_min_freight.show() arrive que l'ensemble est exécuté selon le DAG. Ce processus permet d'optimiser tous les traitements inscrits dans le DAG avant leur exécution.

Les opérations du type select, filter, groupBy sont des transformations non exécutées immédiatement mais intégrées au DAG, les opérations show, count, write sont des actions qui déclenchent l'exécution du DAG.

Sous http://localhost:4040, on peut voir la liste des jobs et des stages. Pour chaque stage, on peut voir le DAG, les métriques associées et la liste des tasks correspondantes.
Un job est lancé sur une action, un job se décompose en plusieurs étapes (stages), la séparation entre stages correspondant principalement à un shuffle. Une étape est exécutée en parallèle sous forme de tasks. Donc un "stage" est un ensemble de "tasks" qui peuvent être exécutées parallèlement sans nécessiter de shuffle entre elles.
